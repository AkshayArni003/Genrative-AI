```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sentance=[
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good'
    ]
```

```python
##define the vocabolory size
voc_size = 10000
```

```python
##One hot representation
one_hot_representation = [one_hot(words, voc_size) for words in sentance]
one_hot_representation
```

```python
##word embedding representation

from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
max_sentence_length = 8
embedded_docs = pad_sequences(one_hot_representation, padding="pre", maxlen=max_sentence_length)
embedded_docs
```

```python
##feature representation
dimension = 10
```

```python
model = Sequential()
model.add(Embedding(voc_size, dimension))
model.build(input_shape=(None, max_sentence_length))
model.compile("adam", "mse")
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

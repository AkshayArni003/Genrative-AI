Facebook AI Similarity Search (Faiss) is a library

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_ollama import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter
loader = TextLoader("speech.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=30)
docs = text_splitter.split_documents(documents)
print(docs)
```

```python
embeddings = OllamaEmbeddings(model="gemma:2b")
db = FAISS.from_documents(docs, embeddings)
db
```

```python
query = "How does the speaker describes the desired outcome of the war?"
docs = db.similarity_search(query)
# docs[1]
```

```python
##Retriver helps to convert the vectore store db into retriver class. This allows to easily use it in other Langchain methods
retriever = db.as_retriever()
docs = retriever.invoke(query)
docs[0].page_content
```

```python
##Similarity Search with score in FAISS similarity_search_with_score method which gives the distance score along with the document. The returned distnace score is L2 distance. Therefore a lower score is better
docs_and_score = db.similarity_search_with_score(query)
docs_and_score
```

```python
##search with vector
embedding_vector = embeddings.embed_query(query)
embedding_vector
```

```python
docs_and_score_by_vector = db.similarity_search_by_vector(embedding_vector)
docs_and_score_by_vector
```

```python
###Saving the db and loading
db.save_local("faiss_index")
```

```python
new_db = FAISS.load_local("faiss_index", embeddings, allow_dangerous_deserialization=True)
docs = new_db.similarity_search(query)
docs
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"] = os.getenv("OPEN_AI_API_KEY")
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_PROJECT"] = os.getenv("LANGCHAIN_PROJECT")
```

```python
from langchain_community.document_loaders import WebBaseLoader
import bs4
```

```python
web_loader = WebBaseLoader(web_paths=("https://docs.langchain.com/oss/python/langchain/messages",))
web_doc = web_loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
final_documents = text_splitter.split_documents(web_doc)
```

```python
from langchain_ollama import OllamaEmbeddings
embeddings = OllamaEmbeddings(model="gemma:2b") ##by default it uses llama2
```

```python
from langchain_community.vectorstores import FAISS
db = FAISS.from_documents(final_documents, embeddings)
db
```
